```python
# Task 2 - Data Cleaning and Preprocessing
# Dataset: Titanic

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load Titanic dataset directly
df = sns.load_dataset('titanic')

print("Raw Dataset Shape")
print(df.shape)

print("\nFirst 5 Rows")
print(df.head())
```

```python
print(" Missing Values ")
print(df.isnull().sum())

print("\nMissing Values % ")
print((df.isnull().sum() / len(df)) * 100)
```

```python
# Fill missing 'age' with median
df['age'] = df['age'].fillna(df['age'].median())

# Fill missing 'embarked' with mode
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

# Fill missing 'embark_town' with mode
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])

# Drop 'deck' column only if it exists
if 'deck' in df.columns:
    df.drop(columns=['deck'], inplace=True)
    print("deck column dropped!")
else:
    print(" deck column already removed, skipping...")

print(" Missing values handled!")
print(df.isnull().sum())
```

```python
print(f"Before removing duplicates: {df.shape}")

df.drop_duplicates(inplace=True)

print(f"After removing duplicates: {df.shape}")
print(" Duplicates removed!")
```

```python
# Check outliers using IQR method

for col in ['age', 'fare']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)].shape[0]
    print(f"{col}: {outliers} outliers found | Range: {lower:.2f} to {upper:.2f}")

    # Cap the outliers (Winsorization)
    df[col] = df[col].clip(lower, upper)

print("\nOutliers handled!")
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

# Normalize age and fare (scale between 0 and 1)
df[['age', 'fare']] = scaler.fit_transform(df[['age', 'fare']])

print(" Features normalized!")
print(df[['age', 'fare']].head())
```

```python
# Convert sex to 0/1
df['sex'] = df['sex'].map({'male': 0, 'female': 1})

# Convert embarked to numbers
df['embarked'] = df['embarked'].map({'S': 0, 'C': 1, 'Q': 2})

print("Categorical columns encoded!")
print(df[['sex', 'embarked']].head())
```

```python
print("Final Dataset Shape ")
print(df.shape)

print("\nData Types ")
print(df.dtypes)

print("\n Final Missing Values Check ")
print(df.isnull().sum())

print("\nSample Clean Data ")
print(df.head())
```

```python
df.to_csv("titanic_cleaned.csv", index=False)
print("Clean dataset saved as titanic_cleaned.csv")
```
